## Deep learning preparation

```python
# Define X and y
X = df
y = df_test
```

```python
# Convert X and y to arrays
X = np.array(X)
y = np.array(y)
```

```python
X
```

```python
y
```

```python
# Reshape X
X = X.reshape(-1,15,9)
```

```python
X.shape
```

```python
y.shape
```

```python
# Split the data to training and testing
X_train, X_test, y_train, y_test = train_test_split(X,y,random_state = 42)
```

## Create model and run RNN

```python
#Create a Keras model
epochs = 30
batch_size = 16
n_hidden = 16

timesteps = len(X_train[0])
input_dim = len(X_train[0][0])
n_classes = len(y_train[0])

model = Sequential()
model.add(Conv1D(n_hidden, kernel_size=3, activation='relu', input_shape=(timesteps, input_dim)))
model.add(MaxPooling1D(pool_size=2))
model.add(LSTM(n_hidden, input_shape=(timesteps, input_dim)))
model.add(Dropout(0.3))
model.add(Dense(n_classes, activation='softmax')) #Don't use relu here!
```

```python
model.compile(loss='categorical_crossentropy',
              optimizer='rmsprop',
              metrics=['accuracy'])
```

```python
model.fit(X_train,
          y_train,
          batch_size=batch_size,
          validation_data=(X_test, y_test),
          epochs=epochs)
```

## Evaluate the model

```python
# Create the weather stations dictionary
stations = {
0: 'BASEL',
1: 'BELGRADE',
2: 'BUDAPEST',
3: 'DEBILT',
4: 'DUSSELDORF',
5: 'HEATHROW',
6: 'KASSEL',
7: 'LJUBLJANA',
8: 'MAASTRICHT',
9: 'MADRID',
10: 'MUNCHENB',
11: 'OSLO',
12: 'SONNBLICK',
13: 'STOCKHOLM',
14: 'VALENTIA'
}
```

```python
# Create confusion matrix for the weather stations

def confusion_matrix(Y_true, Y_pred):
    Y_true = pd.Series([stations[y] for y in np.argmax(Y_true, axis=1)])
    Y_pred = pd.Series([stations[y] for y in np.argmax(Y_pred, axis=1)])

    return pd.crosstab(Y_true, Y_pred, rownames=['True'], colnames=['Pred'])
```

```python
# Evaluate
print(confusion_matrix(y_test, model.predict(X_test)))
```
